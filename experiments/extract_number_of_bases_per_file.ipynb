### executing for all files

```python
dir_with_files = "data"
json_mapping_file = Path(dir_with_files, "file_mapping.json")

with open(json_mapping_file) as f:
    mapping_files = json.load(f)
files = list(mapping_files.keys())
```

```python
dna_bases_count = dict()
for file_name in files:
    logger.info(file_name)
    file_path = Path(dir_with_files, f"{file_name}.fna")
    dna_sequence = []
    counting_bases = {
        "A": 0,
        "C": 0,
        "G": 0,
        "T": 0,
        "N": 0
    }
    has_extra = False
    bases = set(counting_bases.keys())
    with open(file_path, "r") as f:
        content = f.readlines()
        for og_line in content:
            line = og_line.strip().upper()
            if not line.startswith(">"):
                for base in counting_bases.keys():
                    counting_bases[base] += line.count(base)
                if (not has_extra) and (set(line) - bases - {"N"}):
                    has_extra = True

    logger.info(counting_bases)
    if has_extra:
        logger.info("has extra")
    dna_bases_count[file_name] = counting_bases
```
