```python
import kagglehub
from kagglehub import KaggleDatasetAdapter
```

retrieving one domain at a time

```python
dataset_name = "camnugent/genome-information-for-sequenced-organisms"
file_eukaryotes = "eukaryotes.csv"
file_prokaryotes = "prokaryotes.csv"
file_viruses = "viruses.csv"
```

```python
def download_dataset(dataset, file):
    df = kagglehub.dataset_load(
        KaggleDatasetAdapter.PANDAS,
        dataset,
        file,
    )
    return df
```

```python
df_eukaryotes = download_dataset(dataset_name, file_eukaryotes)
df_prokaryotes = download_dataset(dataset_name, file_prokaryotes)
df_viruses = download_dataset(dataset_name, file_viruses)
```
