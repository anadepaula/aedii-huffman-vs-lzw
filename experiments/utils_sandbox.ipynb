## downloading files with requests

```python
import requests
```

```python
filename = "tripanossomussss"
url = (
    "https://ftp.ncbi.nlm.nih.gov/genomes/all/GCA/"
    "000/209/065/GCA_000209065.1_ASM20906v1/"
    "GCA_000209065.1_asm20906v1_genomic.fna.gz"
)
```

```python
response = requests.get(url, stream=True)
```

```python
with open(filename, 'wb') as f:
    f.write(response.content)
```

```python
import urllib
```

```python
urllib.request.urlretrieve(url, filename)
```

```python
import gzip
```

```python
def decompress_gz(input_filepath, output_filepath):
    """
    Decompresses a .gz file.

    Args:
        input_filepath (str): The path to the compressed .gz file.
        output_filepath (str): The path where the decompressed file will be saved.
    """
    with gzip.open(input_filepath, 'rb') as f_in:
        with open(output_filepath, 'wb') as f_out:
            f_out.write(f_in.read())
```

```python
decompress_gz(filename, "a.fna")
```
