## splitting string without consuming delimiter

```python
import re
```

```python
sentence = "the quick brown fox jumps over the (lazy) dog! like, i don't care"
```

```python
re.split(r'([\(\),;\ ])', sentence)
```

```python
re.split(r'([^a-zA-Z0-9])', sentence)
```
